```python
from glob import glob

stocker_indexes_path = "\\".join(str(Path(os.path.dirname(os.getcwd()))).split("\\")[:-2]) + "\\trojan\stockers\index_stocker.csv"

all_sub_indexes = []
indexes_paths   = []
MAIN_paths      = []
for i in dp.getFreshCutIndexes():
    all_sub_indexes.append(i.split("\\")[-1])
    indexes_paths.append(i)
    MAIN_globbed_path = "\\".join(i.split("\\")[:-2]) + "\\MAIN*.csv"
    MAIN_paths.append(glob(MAIN_globbed_path)[0])

stocker_indexes                       = pd.DataFrame()
stocker_indexes["File Name"]          = all_sub_indexes
stocker_indexes["Index"]              = [sub.split("----")[1][:-4] for sub in all_sub_indexes]
stocker_indexes["ID"]                 = [np.nan       for _ in range(len(all_sub_indexes))]
stocker_indexes["Org Name"]           = [np.nan       for _ in range(len(all_sub_indexes))]

stocker_indexes["Patternized"]        = [np.nan       for _ in range(len(all_sub_indexes))]
stocker_indexes["Last Updated"]       = [dp.get_modification_date(p) for p in indexes_paths]

stocker_indexes["MAIN Path"]          = MAIN_paths
stocker_indexes["Path"]               = indexes_paths
stocker_indexes["Execution Duration"] = [np.nan       for _ in range(len(all_sub_indexes))]
dp.overwrite_save_index_stocker(stocker_indexes)
dp.get_index_stocker()
```

# Create new `MAINs stocker` dataframe and csv

```python
root_path = str(Path(os.path.dirname(os.getcwd()))).split("\\")
if root_path[1] == "Users" and root_path[2] == "Amin":
    MAINs_stocker_path = r"/workspace/qgi\trojan\stockers\MAINs_stocker.csv"
else:
    MAINs_stocker_path = r"/workspace/data\trojan\stockers\MAINs_stocker.csv"

MAINs_stocker   = pd.DataFrame()
main_id         = []
main_index_name = []
cutted          = []
subs_path       = []
subs_count      = []
addition_date   = []
org             = []
paths           = []

for MAIN in dp.get_MAINs_to_cut():
    SINGLE_SUBs_path = "\\".join(MAIN.split("\\")[:-1]) + "\\SINGLE_SUBs"

    main_id.append(str(uuid.uuid4()))
    main_index_name.append(MAIN.split("\\")[-1].replace(".csv", "").replace("MAIN", "").replace("-", "").strip())
    if os.path.isdir(SINGLE_SUBs_path):
        cutted.append(True)
        subs_path.append(SINGLE_SUBs_path)
        subs_count.append(len(glob.glob(SINGLE_SUBs_path + "\\*.csv")))
    else:
        cutted.append(False)
        subs_path.append(np.nan)
        subs_count.append(0)
    org.append(MAIN.split("\\")[-2])
    paths.append(MAIN)
    addition_date.append(strftime("%Y-%m-%d %H:%M:%S", gmtime()))
    
MAINs_stocker["id"]               = main_id
MAINs_stocker["MAIN Index Name"]  = main_index_name
MAINs_stocker["Org Name"]         = org
MAINs_stocker["Cutted"]           = cutted
MAINs_stocker["MAIN Path"]        = paths
MAINs_stocker["SINGLE SUBs Path"] = subs_path
MAINs_stocker["SUBs Count"]       = subs_count
MAINs_stocker["Stocked Date"]     = addition_date
MAINs_stocker.to_csv(MAINs_stocker_path)
MAINs_stocker
```
