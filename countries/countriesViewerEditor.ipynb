```python
import pandas as pd
import numpy as np

countries_CSV_path = r"C:\Users\NEW\Desktop\Projects\FIFADash\trojan\countries\countries.csv"
```

```python
def getIndexes(countries, countryName):
    """ 
    Get index positions of value in dataframe. 
    """

    listOfPos = list()
    # Get bool dataframe with True at positions where the given value exists
    result = countries.isin([countryName])
    # Get list of columns that contains the value
    seriesObj = result.any()
    columnNames = list(seriesObj[seriesObj == True].index)
    # Iterate over list of columns and fetch the rows indexes where value exists
    for col in columnNames:
        rows = list(result[col][result[col] == True].index)
        for row in rows:
            listOfPos.append((row, col))
    # Return a list of tuples indicating the positions of value in the dataframe
    return listOfPos

def fixCountryName(countryName):
    countries = pd.read_csv(countries_CSV_path).drop("Unnamed: 0", axis = 1)
    indexes   = getIndexes(countries, countryName)

    if len(indexes) > 0:
        for tuple in indexes:
            return countries.iloc[tuple[0], 0]
```
