## Create new base index stocker dataframe and csv

```python
import sys
sys.path.insert(1, r"/workspace/data\trojan\data engine")
from datetime import datetime
import dataPrep as dp
import pandas   as pd
import numpy    as np
import platform
import os

stocker_indexes_path = r"/workspace/data\trojan\stockers\index_stocker.csv"

all_sub_indexes = []
indexes_paths   = []
for i in dp.getFreshCutIndexes():
    all_sub_indexes.append(i.split("\\")[-1])
    indexes_paths.append(i)

stocker_indexes                 = pd.DataFrame()
stocker_indexes["Index"]        = all_sub_indexes
stocker_indexes["Patternized"]  = [np.nan for _ in range(len(all_sub_indexes))]
stocker_indexes["Last Updated"] = [dp.get_modification_date(p) for p in indexes_paths]
stocker_indexes["Path"]         = indexes_paths
stocker_indexes.to_csv(stocker_indexes_path)


df = pd.read_csv(stocker_indexes_path).drop("Unnamed: 0", axis = 1)
df.head()
```
