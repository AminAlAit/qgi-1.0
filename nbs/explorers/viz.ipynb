```python
variation = 12


dfViz              = dp.patternVsMatcheDF(sortedPatternsDict[list(sortedPatternsDict.keys())[variation]])
country1           = ''.join([i for i in list(sortedPatternsDict.keys())[variation].split("-")[0] if not i.isdigit()])
country2           = ''.join([i for i in list(sortedPatternsDict.keys())[variation].split("-")[2] if not i.isdigit()])
patternLen         = int(list(sortedPatternsDict.keys())[variation].split("-")[1])
country1IndexStart = list(sortedPatternsDict[list(sortedPatternsDict.keys())[variation]]["Pattern Index Start"])[0]
country1IndexEnd   = list(sortedPatternsDict[list(sortedPatternsDict.keys())[variation]]["Pattern Index End"])[0]
country2IndexStart = list(sortedPatternsDict[list(sortedPatternsDict.keys())[variation]]["Match Index Start"])[0]
country2IndexEnd   = list(sortedPatternsDict[list(sortedPatternsDict.keys())[variation]]["Match Index End"])[0]
corrScores         = list(sortedPatternsDict[list(sortedPatternsDict.keys())[variation]]["Correlation Score"])

country1Cols = []
country2Cols = []
for col in list(dfViz):
    if country1 in col:
        country1Cols.append(col)
    elif country2 in col:
        country2Cols.append(col)

## Viz part
mainTitle = """<H1><b>Trojan Geopolitics</b></H1>"""
mainTitle += """\n<H2><FONT COLOR = "#0000FF">""" + country1 + """ (""" + str(country1IndexStart) + """ - """ + str(country1IndexEnd) + """)</FONT>""" + " vs. "
mainTitle += """<FONT COLOR = "#FFA500">"""       + country2 + """ (""" + str(country2IndexStart) + """ - """ + str(country2IndexEnd) + """)</FONT>, with a pattern length of """
mainTitle += str(patternLen) + """ years. </H2>"""
mainTitle += """<H3>\nPattern strength: """ + str(len(country1Cols)) + """ indexes, with average correlation of """ + str(round(np.mean(corrScores), 2) * 100) + """%.</H3> """

plots = []
for i in range(len(country1Cols)):
    country1Col = country1Cols[i]
    country2Col = country2Cols[i]
   
    fig   = figs(width = 150, height = 100, title = country1Col.split("-")[-1] + "\n(r = " + str(round(corrScores[i] * 100, 2)) + "%)")
    years = list(dfViz["Years"])
    
    # Plotting country 1  
    fig.line(
        years,
        list(dfViz[country1Col]),
        line_color   = "blue")
        #line_dash    = "solid",
        #legend_label = country1Col.split("-")[0])
    
    # Plotting country 2
    fig.line(
        years,
        list(dfViz[country2Col]),
        line_color       = "orange")
        #line_dash        = "dotdash",
        #line_dash_offset = 1,
        #legend_label     = country2Col.split("-")[0])
    
    plots.append(fig)

titleDiv = Div(text = mainTitle, width = 1800, height = 120)

show(column(titleDiv, gridplot(plots, ncols = 3, width = 300, height = 250)))

dfViz
```
